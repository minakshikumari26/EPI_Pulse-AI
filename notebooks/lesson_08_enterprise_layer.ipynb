## What Each Enterprise Folder Does

- `api/routes/`: separates endpoint logic from the FastAPI app setup.
- `api/schemas/`: validates API input and output with Pydantic.
- `database/`: defines SQLAlchemy sessions and table models.
- `docker/`: contains separate Dockerfiles for API and dashboard.
- `kafka/`: provides event producer and consumer scripts.
- `airflow/`: defines a daily outbreak pipeline DAG.
- `mlflow_tracking/`: logs experiment metrics.
- `k8s/`: contains Kubernetes deployment and service manifests.
- `monitoring/`: exposes a logging entrypoint.
