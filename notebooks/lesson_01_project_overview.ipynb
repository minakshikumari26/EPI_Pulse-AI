# Lesson 1: Project Overview

EpiPulse AI is a prototype epidemic monitoring system. It tracks disease cases, detects outbreak spikes, estimates regional risk, generates short-term forecasts, and exposes results through a dashboard and API.

## Project Folders

- `data/`: disease case dataset
- `src/preprocessing/`: cleaning and feature engineering
- `src/anomaly_detection/`: spike and anomaly detection
- `src/forecasting/`: ARIMA and Prophet forecasting
- `src/risk_scoring/`: regional risk scores
- `src/alerting/`: outbreak alert messages
- `src/geospatial/`: heatmap generation
- `dashboard/`: Streamlit dashboard
- `api/`: FastAPI service
- `configs/`: central settings
- `tests/`: automated checks
