# Lesson 6: Dashboard and API

This lesson explains how to run the Streamlit dashboard and FastAPI service.

## Dashboard

Run this command from the project root:

```powershell
streamlit run dashboard/app.py
```

Expected result: a local dashboard URL, usually `http://localhost:8501`.

The dashboard shows KPI cards, region filters, case trends, spike markers, risk comparison, 7-day projection, report, conclusion, and raw data inside an expander.

## API

Run this command from the project root:

```powershell
uvicorn api.main:app --reload
```

Open `http://localhost:8000/docs` for interactive API documentation.
